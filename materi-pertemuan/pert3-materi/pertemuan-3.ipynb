## NumPy vs Pandas: Buat Apa Masing-Masing di Dunia Data Analyst?

Sebelum masuk ke kode, kenalan dulu sama dua library yang dipakai sepanjang pertemuan ini.

**NumPy (Numerical Python)**
- Library dasar buat komputasi numerik di Python — isinya struktur data **array** dan operasi matematis yang cepat.
- Dipakai data analyst buat hitung-hitungan angka dalam jumlah besar sekaligus (operasi vectorized), tanpa nulis loop manual satu-satu.
- Array NumPy isinya harus **satu tipe data yang sama** (semua angka, misalnya), dan tidak punya nama kolom/baris — cuma deretan angka polos.

**Pandas**
- Library buat mengolah **data tabular** (tabel) — dibangun di atas NumPy, jadi performanya tetap cepat, tapi lebih ramah dipakai buat data beneran.
- Dipakai data analyst buat baca file (CSV, Excel), bersihkan data, filter, gabungkan tabel, sampai ringkas jadi insight.
- Struktur datanya (Series, DataFrame) boleh campur tipe data per kolom (satu kolom teks, kolom lain angka, kolom lain tanggal), dan punya nama kolom + index — persis seperti tabel Excel.

**Perbedaannya secara singkat:**

| | NumPy | Pandas |
|---|---|---|
| Struktur data | Array (angka polos, satu tipe) | Series/DataFrame (tabel, boleh campur tipe) |
| Cocok buat | Komputasi angka murni, matematika/statistik dasar | Data terstruktur: baca CSV, bersihkan, filter, analisis |
| Ada nama kolom? | Tidak | Ya (dan index baris) |
| Dipakai kapan | Di balik layar Pandas, atau perhitungan angka murni | Sehari-hari buat kerja dengan dataset asli |

Intinya: **NumPy itu mesinnya, Pandas itu yang dipakai langsung.** Data analyst jarang menyentuh NumPy secara langsung untuk data tabular — tapi tiap kali menjalankan operasi Pandas (`df["harga"] * 2`, `df["harga"].mean()`, dst), di baliknya NumPy yang bekerja supaya prosesnya cepat.

## 9. Statistik Satu Angka

```python
print("Harga termurah:", df["harga"].min())
print("Harga termahal:", df["harga"].max())
print("Rata-rata harga:", df["harga"].mean())
print("Median harga:", df["harga"].median())
print("Kuartil 1 (Q1):", df["harga"].quantile(0.25))
print("Kuartil 3 (Q3):", df["harga"].quantile(0.75))
print("Total jumlah barang terjual:", df["jumlah"].sum())
```

Sekarang cari **produk**-nya, bukan cuma angkanya.

```python
produk_termahal = df.loc[df["harga"].idxmax()]
produk_termurah = df.loc[df["harga"].idxmin()]

print("Produk termahal:")
print(produk_termahal[["nama_produk", "kategori", "harga"]])

print("\nProduk termurah:")
print(produk_termurah[["nama_produk", "kategori", "harga"]])
```

## 15. Feature Engineering: Kolom nilai_penjualan

`nilai_penjualan` belum ada di `data.csv` — dibuat dari kolom `harga` dan `jumlah` yang sudah ada.

```python
df["nilai_penjualan"] = df["harga"] * df["jumlah"]
display(df[["nama_produk", "harga", "jumlah", "nilai_penjualan"]].head())
```

## 16. Menghitung Untung/Rugi per Transaksi

```python
df["untung"] = (df["harga"] - df["modal"]) * df["jumlah"]
display(df[["nama_produk", "harga", "modal", "jumlah", "untung"]].head())

produk_paling_untung = df.loc[df["untung"].idxmax()]
produk_paling_rugi = df.loc[df["untung"].idxmin()]

print("Produk paling untung:")
print(produk_paling_untung[["nama_produk", "kategori", "harga", "modal", "jumlah", "untung"]])

print("\nProduk paling rugi:")
print(produk_paling_rugi[["nama_produk", "kategori", "harga", "modal", "jumlah", "untung"]])

print("\nJumlah baris untung:", (df["untung"] > 0).sum())
print("Jumlah baris rugi:", (df["untung"] < 0).sum())
```

## 18. (Bonus) Menggabungkan Dua Tabel dengan merge()

```python
diskon_kategori = pd.DataFrame({
    "kategori": ["Elektronik", "Fashion", "Makanan", "Olahraga"],
    "diskon_persen": [10, 15, 5, 20],
})
display(diskon_kategori)

df_inner = pd.merge(df, diskon_kategori, on="kategori", how="inner")
print("INNER JOIN shape:", df_inner.shape)

df_gabung = pd.merge(df, diskon_kategori, on="kategori", how="left")
print("LEFT JOIN shape:", df_gabung.shape)
display(df_gabung[df_gabung["kategori"] == "Buku"][["nama_produk", "kategori", "diskon_persen"]].head(3))
```

## 19. (Bonus) Menyimpan Hasil ke File CSV Baru

```python
df_gabung.to_csv("data_dengan_diskon.csv", index=False)
print("Tersimpan sebagai data_dengan_diskon.csv")
```
