## NumPy vs Pandas: Buat Apa Masing-Masing di Dunia Data Analyst?

Sebelum masuk ke kode, kenalan dulu sama dua library yang dipakai sepanjang pertemuan ini.

**NumPy (Numerical Python)**
- Library dasar buat komputasi numerik di Python — isinya struktur data **array** dan operasi matematis yang cepat.
- Dipakai data analyst buat hitung-hitungan angka dalam jumlah besar sekaligus (operasi vectorized), tanpa nulis loop manual satu-satu.
- Array NumPy isinya harus **satu tipe data yang sama** (semua angka, misalnya), dan tidak punya nama kolom/baris — cuma deretan angka polos.

**Pandas**
- Library buat mengolah **data tabular** (tabel) — dibangun di atas NumPy, jadi performanya tetap cepat, tapi lebih ramah dipakai buat data beneran.
- Dipakai data analyst buat baca file (CSV, Excel), bersihkan data, filter, gabungkan tabel, sampai ringkas jadi insight.
- Struktur datanya (Series, DataFrame) boleh campur tipe data per kolom (satu kolom teks, kolom lain angka, kolom lain tanggal), dan punya nama kolom + index — persis seperti tabel Excel.

**Perbedaannya secara singkat:**

| | NumPy | Pandas |
|---|---|---|
| Struktur data | Array (angka polos, satu tipe) | Series/DataFrame (tabel, boleh campur tipe) |
| Cocok buat | Komputasi angka murni, matematika/statistik dasar | Data terstruktur: baca CSV, bersihkan, filter, analisis |
| Ada nama kolom? | Tidak | Ya (dan index baris) |
| Dipakai kapan | Di balik layar Pandas, atau perhitungan angka murni | Sehari-hari buat kerja dengan dataset asli |

Intinya: **NumPy itu mesinnya, Pandas itu yang dipakai langsung.** Data analyst jarang menyentuh NumPy secara langsung untuk data tabular — tapi tiap kali menjalankan operasi Pandas (`df["harga"] * 2`, `df["harga"].mean()`, dst), di baliknya NumPy yang bekerja supaya prosesnya cepat.

## 9. Statistik Satu Angka

## 14. Filter Baris: Harga di Atas 100.000

```python
produk_mahal = df[df["harga"] > 100000]
print(produk_mahal.shape)
display(produk_mahal.head())
```

Satu kondisi doang kadang belum cukup — coba gabungan kondisi (AND/OR/NOT).

```python
# AND (&) -- harga di atas 100rb DAN kategorinya Elektronik
elektronik_mahal = df[(df["harga"] > 100000) & (df["kategori"] == "Elektronik")]
print("Elektronik mahal:", elektronik_mahal.shape)
display(elektronik_mahal[["nama_produk", "kategori", "harga"]].head())

# Satu kondisi lain -- semua produk yang terjual di wilayah tertentu
produk_jakarta = df[df["wilayah"] == "Jakarta"]
print("Produk di Jakarta:", produk_jakarta.shape)
display(produk_jakarta[["nama_produk", "wilayah"]].head())

# NOT (!= atau ~) -- semua produk SELAIN kategori Buku
bukan_buku = df[df["kategori"] != "Buku"]
print("Bukan kategori Buku:", bukan_buku.shape)
display(bukan_buku[["nama_produk", "kategori"]].head())
```

## 15. Feature Engineering: Kolom nilai_penjualan

`nilai_penjualan` belum ada di `data.csv` — dibuat dari kolom `harga` dan `jumlah` yang sudah ada.

```python
df["nilai_penjualan"] = df["harga"] * df["jumlah"]
display(df[["nama_produk", "harga", "jumlah", "nilai_penjualan"]].head())
```

## 16. Menghitung Untung/Rugi per Transaksi

```python
df["untung"] = (df["harga"] - df["modal"]) * df["jumlah"]
display(df[["nama_produk", "harga", "modal", "jumlah", "untung"]].head())

produk_paling_untung = df.loc[df["untung"].idxmax()]
produk_paling_rugi = df.loc[df["untung"].idxmin()]

print("Produk paling untung:")
display(produk_paling_untung[["nama_produk", "kategori", "harga", "modal", "jumlah", "untung"]])

print("\nProduk paling rugi:")
display(produk_paling_rugi[["nama_produk", "kategori", "harga", "modal", "jumlah", "untung"]])

print("\nJumlah baris untung:", (df["untung"] > 0).sum())
print("Jumlah baris rugi:", (df["untung"] < 0).sum())
```

## 18. (Bonus) Menggabungkan Dua Tabel dengan merge()

Coba dulu di tabel kecil biar keempat jenis join-nya keliatan jelas.

```python
pelanggan = pd.DataFrame({
    "id_pelanggan": [1, 2, 3],
    "nama": ["Andi", "Budi", "Citra"],
})

pesanan = pd.DataFrame({
    "id_pelanggan": [1, 2, 4],
    "produk": ["Mouse", "Kaos", "Novel"],
})

display(pelanggan)
display(pesanan)

for cara in ["inner", "left", "right", "outer"]:
    print(f"\n--- how={cara} ---")
    display(pd.merge(pelanggan, pesanan, on="id_pelanggan", how=cara))
```

Sekarang terapkan ke data asli -- tapi dites dulu di potongan kecil (4 baris) biar hasil tiap join masih kebaca penuh sebagai tabel, sebelum diterapkan ke semua 133 baris.

```python
diskon_kategori = pd.DataFrame({
    "kategori": ["Elektronik", "Fashion", "Makanan", "Olahraga"],
    "diskon_persen": [10, 15, 5, 20],
})
display(diskon_kategori)

# Contoh kecil: 2 produk Elektronik (kategorinya ADA di diskon_kategori)
# + 2 produk Buku (kategorinya TIDAK ADA di diskon_kategori)
bersih = df[df["harga"].notna()]
contoh = bersih[bersih["kategori"].isin(["Elektronik", "Buku"])].head(4)[["nama_produk", "kategori", "harga"]]
display(contoh)

for cara in ["inner", "left", "right", "outer"]:
    print(f"\n--- how={cara} ---")
    display(pd.merge(contoh, diskon_kategori, on="kategori", how=cara))
```

Sekarang terapkan ke keseluruhan data (133 baris) -- cukup lihat ukurannya, tabelnya kepanjangan buat ditampilkan penuh.

```python
for cara in ["inner", "left", "right", "outer"]:
    hasil = pd.merge(df, diskon_kategori, on="kategori", how=cara)
    print(f"{cara.upper()} JOIN shape:", hasil.shape)

df_gabung = pd.merge(df, diskon_kategori, on="kategori", how="left")
display(df_gabung[df_gabung["kategori"] == "Buku"][["nama_produk", "kategori", "diskon_persen"]].head(3))
```

## Cheatsheet: NumPy & Pandas

Ringkasan semua perintah yang dipakai hari ini, biar gampang dicari lagi pas lupa.

### NumPy

| Perintah | Buat apa |
|---|---|
| `np.array([...])` | Bikin array dari list biasa |
| `array * 2` | Operasi vectorized -- semua elemen dikali sekaligus, tanpa loop |

### Pandas: Bikin & Baca Data

| Perintah | Buat apa |
|---|---|
| `pd.Series([...], name="...")` | Bikin satu kolom data berlabel |
| `pd.DataFrame({...})` | Bikin tabel dari dictionary |
| `pd.read_csv("nama_file.csv")` | Baca file CSV jadi DataFrame |
| `df.to_csv("nama_file.csv", index=False)` | Simpan DataFrame jadi file CSV baru |

### Pandas: Mengintip & Memahami Data (Data Understanding)

| Perintah | Buat apa |
|---|---|
| `df.head()` | Lihat 5 baris pertama (default) |
| `df.tail(n)` | Lihat `n` baris terakhir |
| `df.shape` | `(jumlah baris, jumlah kolom)` |
| `df.info()` | Struktur: nama kolom, tipe data, jumlah non-null |
| `df.describe()` | Statistik cepat (count/mean/std/min/25%/50%/75%/max) buat kolom angka |
| `df.isna().sum()` | Jumlah data kosong (`NaN`) per kolom |
| `df.duplicated().sum()` | Jumlah baris yang duplikat (cuma dihitung, bukan baris aslinya) |
| `df.duplicated(keep=False)` | Tandai SEMUA baris yang punya kembaran (asli + salinan) |

### Pandas: Statistik Satu Kolom

| Perintah | Buat apa |
|---|---|
| `df["kolom"].min()` / `.max()` | Nilai terkecil / terbesar |
| `df["kolom"].mean()` / `.median()` | Rata-rata / nilai tengah |
| `df["kolom"].quantile(0.25)` / `.quantile(0.75)` | Kuartil 1 (Q1) / Kuartil 3 (Q3) |
| `df["kolom"].sum()` | Total semua nilai di kolom |
| `df["kolom"].idxmax()` / `.idxmin()` | Index baris dengan nilai terbesar / terkecil |
| `df.loc[index]` | Ambil satu baris lengkap berdasarkan index-nya |

### Pandas: Memilih & Menyaring Data

| Perintah | Buat apa |
|---|---|
| `df["kolom"]` | Ambil satu kolom -> Series |
| `df[["a", "b"]]` | Ambil beberapa kolom sekaligus -> DataFrame |
| `df[df["kolom"] > angka]` | Filter baris berdasarkan satu kondisi |
| `&` | AND -- gabung 2+ kondisi, semuanya harus benar (wajib dikurung tiap kondisi) |
| `\|` | OR -- gabung 2+ kondisi, salah satu benar sudah cukup |
| `~` atau `!=` | NOT -- membalik/menegasikan kondisi |
| `df["baru"] = df["a"] * df["b"]` | Bikin kolom baru dari kolom yang sudah ada (feature engineering) |
| `df.select_dtypes(include="number")` | Ambil cuma kolom-kolom yang bertipe angka |

### Pandas: Menggabungkan Tabel (Join)

| Perintah | Buat apa |
|---|---|
| `pd.merge(a, b, on="kolom", how="inner")` | Cuma baris yang kuncinya ada di KEDUA tabel |
| `how="left"` | Semua baris tabel kiri dipertahankan, yang tidak match diisi `NaN` |
| `how="right"` | Semua baris tabel kanan dipertahankan, yang tidak match diisi `NaN` |
| `how="outer"` | Semua baris dari kedua tabel, yang tidak match diisi `NaN` |
